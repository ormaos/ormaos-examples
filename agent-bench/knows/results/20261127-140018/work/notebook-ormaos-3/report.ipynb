# Weekly sales, with tax

The sales of the week from `sales.csv`: single sales above 1000 are handled by the key-account team and are left out here, then 20% tax is added and the week is summed.

```python
import csv

import ormaos

LIMIT = 1000
TAX = 0.20


def load(path):
    with open(path, newline="") as source:
        return [
            {"sku": row["sku"], "price": float(row["price"]), "quantity": int(row["quantity"])}
            for row in csv.DictReader(source)
        ]


def drop_key_accounts(rows):
    return [row for row in rows if row["price"] <= LIMIT]


def add_tax(rows):
    return [{**row, "price": round(row["price"] * (1 + TAX), 2)} for row in rows]


def total(rows):
    return sum(row["price"] * row["quantity"] for row in rows)
```

```python
with ormaos.record():
    rows = load("sales.csv")
len(rows)
```

```python
with ormaos.record():
    rows = drop_key_accounts(rows)
```

```python
with ormaos.record():
    rows = add_tax(rows)
```

```python
with ormaos.record():
    week = total(rows)
print(f"week total with tax: {week:.2f}")
```
